```python
merged["food_name"] = merged["food_name"].str.lower()

healthy_counts = {}
unhealthy_counts = {}

for seqn, foods in merged.groupby("SEQN")["food_name"]:

    food_text = " ".join(foods.astype(str))

    healthy_counts[seqn] = sum(
        word in food_text for word in healthy_words
    )

    unhealthy_counts[seqn] = sum(
        word in food_text for word in unhealthy_words
    )
```
